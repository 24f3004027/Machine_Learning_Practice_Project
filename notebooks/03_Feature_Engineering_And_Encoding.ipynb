In this milestone, we engineer domain-specific temporal interaction features (Machine Age, Usage Intensity per Year, Depreciation Ratios) and apply out-of-fold safe frequency encoding for high-cardinality categorical variables.

```python
import pandas as pd
import numpy as np

def engineer_domain_features(df):
    df = df.copy()
    if 'YearMade' in df.columns and 'saledate' in df.columns:
        df['saledate'] = pd.to_datetime(df['saledate'])
        df['SaleYear'] = df['saledate'].dt.year
        df['MachineAge'] = df['SaleYear'] - df['YearMade']
        df['MachineAge'] = df['MachineAge'].apply(lambda x: x if 0 <= x <= 80 else np.nan)
    
    if 'MachineHoursCurrentMeter' in df.columns and 'MachineAge' in df.columns:
        df['HoursPerYear'] = df['MachineHoursCurrentMeter'] / (df['MachineAge'] + 1.0)
    return df

print("Milestone 3: Feature Engineering module defined.")
```
