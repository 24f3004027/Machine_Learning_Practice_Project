In this milestone, we implement custom regular expression extractors to parse unformatted text strings into continuous numeric features (Horsepower, Operating Tonnage, Bucket Yardage) and establish baseline Random Forest benchmark.

```python
import re
import pandas as pd
import numpy as np

def extract_hp(desc):
    if not isinstance(desc, str):
        return np.nan
    match = re.search(r'(\d+(?:\.\d+)?)\s*(?:hp|horsepower)', desc, re.IGNORECASE)
    return float(match.group(1)) if match else np.nan

def extract_tonnage(desc):
    if not isinstance(desc, str):
        return np.nan
    match = re.search(r'(\d+(?:\.\d+)?)\s*(?:ton|tonnage)', desc, re.IGNORECASE)
    return float(match.group(1)) if match else np.nan

test_desc = "Caterpillar 330D Excavator 250 HP 32 Ton Capacity"
print(f"Extracted HP: {extract_hp(test_desc)}")
print(f"Extracted Tonnage: {extract_tonnage(test_desc)}")
```
