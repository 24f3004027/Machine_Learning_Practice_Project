In this milestone, we conduct comprehensive EDA on heavy equipment auction data, analyze missingness, examine price skews, and apply logarithmic transformations to stabilize target variance.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Milestone 1: EDA & Target Distribution Environment Ready.")
```

```python
# Target Variable Log-Transformation
def analyze_target_distribution(saleprice_array):
    log_prices = np.log1p(saleprice_array)
    skewness = pd.Series(log_prices).skew()
    print(f"Target Log1p Skewness: {skewness:.4f}")
    return log_prices

sample_prices = np.array([12000, 45000, 89000, 150000, 23000, 67000])
analyze_target_distribution(sample_prices)
```
