## Milestone 5: 5-Seed Ensemble Blending & Jensen's Inequality Correction
**IIT Madras Machine Learning Practice (MLP) Project (S-Grade / 10.0 GPA)**

In this final milestone, we execute a 5-seed weighted log-space ensemble blending XGBoost (45%), LightGBM (35%), and CatBoost (20%), and apply the empirical Jensen's multiplier (1.001300) to achieve the final production RMSLE of 0.18663.

```python
from src.evaluate import blend_predictions, apply_jensens_multiplier, calculate_rmsle
import numpy as np

# Demonstration of Log-Space Blending + Jensen's Multiplier
y_true = np.array([25000, 48000, 95000, 140000])
pred_xgb = np.array([24800, 48500, 94000, 141000])
pred_lgb = np.array([25200, 47800, 95500, 139500])
pred_cat = np.array([24900, 48200, 94800, 140500])

blended = blend_predictions(pred_xgb, pred_lgb, pred_cat, weights=(0.45, 0.35, 0.20))
final_preds = apply_jensens_multiplier(blended, multiplier=1.001300)

rmsle = calculate_rmsle(y_true, final_preds)
print(f"Milestone 5 Production Validation RMSLE: {rmsle:.5f}")
```
